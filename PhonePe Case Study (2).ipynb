This case study is about analyzing digital payment data from PhonePe across Indian states and districts. The goal is to find trends in transactions, user behavior, device usage, and demographics using Python.

Here is a detailed gist of each dataset:

1. State_Txn and Users:
Contains state-level transaction and user data over time. It shows:
a. Number of transactions
b. Total transaction amount (INR)
c. Average transaction value (ATV)
d. Number of registered users and app opens
Purpose of the dataset: Track digital payment adoption and usage at the state level.

2. State_TxnSplit:
Breaks down state-level data by transaction type (e.g., recharge, bill pay, UPI transfer). It provides:
a. Number and amount of transactions for each type
b. Average transaction value per type
Purpose of the dataset: Understand the usage of different payment services in each state.


3. State_DeviceData:
Shows which mobile device brands are used by registered users in each state, including:
a. Brand name
b. Number of users per brand
c. Percentage share per brand
Purpose of the dataset: Analyze device adoption and market penetration of brands.

4. District_Txn and Users:
Provides transaction and user data at a more granular district level, including:
a. Transactions, b. amount, and c. ATV
d. Registered users and e. app opens per district
Purposeof the dataset:Explore digital payment patterns within states by district.

5. District Demographics:
Contains demographic features of each district:
a. Population, b. area, c. population density, d. headquarters, e. code
f. Alternate district names
Purpose of the dataset: Correlate transaction trends with demographics (e.g., high population = more transactions).

By analyzing these datasets together, you can link usage trends to location, population, and device type, giving a 360-degree view of how digital payments are adopted across regions and communities.

Visualization of Distribution of different transaction types of each state for recent quarter :-

Summary of Key Findings Insights :-

The analysis of the PhonePe dataset (covering state-level and district-level metrics from Q1 2018 to Q2 2021) reveals several critical trends in transaction volume, payment categories, user adoption, and device preferences across India :-

Regional Transaction Distribution Market Penetration :-

Top Performing States: Karnataka leads India in total transaction volume, followed closely by Maharashtra,Telangana,Andhra Pradesh and Rajasthan. Metropolitan tech centers like Bengaluru and Hyderabad serve as major adoption engines.
Lowest Volume Regions: Lakshadweep, Andaman & Nicobar Island, Ladakh, Mizoram, and Meghalaya record the lowest transaction volumes, primarily due to smaller population bases.
User-to-Population Ratio:High urban density and strong digital infrastructure drive the highest user-to-population ratios in Delhi, Chandigarh, Telangana, Dadra & Nagar Haveli, and Karnataka. Conversely, states like Meghalaya, Lakshadweep, Mizoram, Nagaland, and Assam show low user penetration relative to their population.

Transaction Types Use Cases :-

Gateway Transactions: In early adoption stages (e.g., Andhra Pradesh in Q1 2018), Recharge & Bill Payments accounted for 55.6% of all transactions, while Peer-to-Peer payments comprised 33.7%. Recurring utility bill payments served as the primary entry point for new users.

Emerging Commercial Payments: While Merchant Payments and Financial Services represented a small percentage during early rollout, transaction types across all states showed steady functional diversification by Q2 2021.

Average Transaction Value (ATV) and Per-User Spending Analysis :-

Highest ATV Regions: Remote islands, and northeastern regions recorded the highest average transaction values per single transaction like Ladakh, Andaman & Nicobar Islands, Mizoram, Lakshadweep and Arunachal Pradesh.

Lowest ATV Regions: Densely populated and highly active digital markets exhibited lower average transaction values per payment like West Bengal, Odisha, Dadra & Nagar Haveli and Daman & Diu, Maharashtra, and Karnataka due to frequent low-ticket daily usage.

Highest Cumulative Spend Per User:Telangana leads in cumulative transaction amount per registered user, followed by Andhra Pradesh,Karnataka, Rajasthan, and Delhi.

 Growth Trajectory & Resilience (West Bengal Case Study) :-

Accelerated Growth: App opens in West Bengal surged from Q2 2019 onward, reaching over 30 crore by Q1.
Lockdown Resilience: During Q2 2020 (COVID-19 lockdown), transaction volume experienced a temporary dip, but average monetary spend per transaction increased, demonstrating payment reliance during crisis periods. Volume and overall value surged steeply from Q3 2020 through Q2 2021.

Smartphone Brand Adoption and Demographic Factors :-

Device Dominance: Xiaomi is the most popular smartphone brand among PhonePe registered users across a majority of states (e.g., Maharashtra with 9.76M registered Xiaomi users, Uttar Pradesh with 8.04M, and Karnataka with 6.57M), followed by Vivo and Samsung. Motorola and Micromax show the lowest adoption ratios.

Population Density Impact: Pearson correlation between a state's population density and its overall transaction volume is near zero (r = 0.048). This indicates that population density alone does not drive digital adoption; factors such as merchant onboarding, smartphone penetration, income levels, and financial literacy play a far bigger role.
